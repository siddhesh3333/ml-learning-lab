```python
documents = {
    0: "The Eiffel Tower is located in Paris, France.",
    1: "Paris is the capital city of France.",
    2: "The French Revolution began in 1789.",
    3: "French cuisine is renowned worldwide for its pastries and wine.",
    4: "The Seine River flows through the heart of Paris.",
    5: "Victor Hugo wrote the novel Les Miserables.",
    6: "The Eiffel Tower was completed in 1889 and was the tallest man-made structure at the time.",
}
```

```python
queries = [
    {"relevant": {0, 6},    "retrieved": [0, 2, 6]},
    {"relevant": {3},       "retrieved": [1, 3, 5]},
    {"relevant": {0, 1, 4}, "retrieved": [0, 1, 4]},
]
```

```python
def precision_at_k(retrieved, relevant, k):
    # Take only the first k retrieved documents.
    top_k = retrieved[:k]

    # Count how many of those top-k documents are actually relevant.
    relevant_count = sum(1 for doc_id in top_k if doc_id in relevant)

    # Precision = relevant retrieved / total retrieved in top-k.
    # Rounded to 2 decimal places as requested.
    return round(relevant_count / k, 2) if k > 0 else 0.0
```

```python

def recall_at_k(retrieved, relevant, k):
    # Take only the first k retrieved documents.
    top_k = retrieved[:k]

    # Count how many of those top-k documents are actually relevant.
    relevant_count = sum(1 for doc_id in top_k if doc_id in relevant)

    # Recall = relevant retrieved / total relevant documents.
    # Rounded to 2 decimal places as requested.
    return round(relevant_count / len(relevant), 2) if relevant else 0.0
```

```python
def reciprocal_rank(retrieved, relevant):
    # Check the retrieved documents one by one in rank order.
    # rank starts at 1 because the first result is rank 1, not 0.
    for rank, doc_id in enumerate(retrieved, start=1):
        # As soon as we find the first relevant document, return 1/rank.
        if doc_id in relevant:
            return round(1 / rank, 2)

    # If no relevant document is found, reciprocal rank is 0.
    return 0.0
```

```python
def mean_reciprocal_rank(all_retrieved, all_relevant):
    # Compute reciprocal rank for each query.
    rr_values = [
        reciprocal_rank(retrieved, relevant)
        for retrieved, relevant in zip(all_retrieved, all_relevant)
    ]

    # Average the reciprocal ranks across all queries.
    # Rounded to 3 decimal places as requested.
    return round(sum(rr_values) / len(rr_values), 3) if rr_values else 0.0
```

```python
if __name__ == "__main__":
    k = 3

    # Collect retrieved lists and relevant sets separately from the queries list.
    all_retrieved = [q["retrieved"] for q in queries]
    all_relevant = [q["relevant"] for q in queries]

    # Compute Precision@3 for each query.
    precision_scores = [
        precision_at_k(q["retrieved"], q["relevant"], k) for q in queries
    ]

    # Compute Recall@3 for each query.
    recall_scores = [
        recall_at_k(q["retrieved"], q["relevant"], k) for q in queries
    ]

    # Compute Reciprocal Rank for each query.
    rr_scores = [
        reciprocal_rank(q["retrieved"], q["relevant"]) for q in queries
    ]

    # Compute Mean Reciprocal Rank across all queries.
    mrr = mean_reciprocal_rank(all_retrieved, all_relevant)

    # Print the final report.
    print("Precision@3:", precision_scores)
    print("Recall:", recall_scores)
    print("Reciprocal Ranks:", rr_scores)
    print("MRR:", mrr)
```
