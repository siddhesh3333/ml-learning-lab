```python
task_examples = [
    {"question": "What is the capital of France?",
     "choices": ["A) London", "B) Paris", "C) Berlin", "D) Madrid"],
     "answer": "B"},
    {"question": "What do plants produce during photosynthesis?",
     "choices": ["A) Oxygen", "B) Nitrogen", "C) Carbon monoxide", "D) Methane"],
     "answer": "A"},
    {"question": "What is 15% of 200?",
     "choices": ["A) 20", "B) 25", "C) 30", "D) 35"],
     "answer": "C"},
    {"question": "Who wrote Romeo and Juliet?",
     "choices": ["A) Charles Dickens", "B) William Shakespeare", "C) Mark Twain", "D) Leo Tolstoy"],
     "answer": "B"},
    {"question": "What is the powerhouse of the cell?",
     "choices": ["A) Nucleus", "B) Ribosome", "C) Mitochondria", "D) Golgi apparatus"],
     "answer": "C"},
]

raw_responses = [
    "B",
    "A, plants release oxygen during photosynthesis",
    "C) 30",
    "The answer is D",
    "C",
]

def  extract_letter(raw_response):
  for ch in  raw_response.upper():
    if ch  in ["A","B","C","D"]:
      return ch
    return None

```

```python
def evaluate_from_responses(task_examples,raw_responses):
  predictions = []
  labels = []

  for example, response in zip(task_examples,raw_responses):
    predicted_letter = extract_letter(response)
    predictions.append(predicted_letter)

    label =  example['answer']
    labels.append(label)


  correct = 0

  for pred ,label  in zip(predictions,labels):
    if pred == label:
      correct +=1

  total = len(labels)
  accuracy = (correct/total )  * 100  if total  > 0  else 0


  return accuracy,total , correct




if __name__ =="__main__":
  print(evaluate_from_responses(task_examples,raw_responses))
```
